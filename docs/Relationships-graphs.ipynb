## Reliability distributions relationships

```raku
my @edges =
    {
        from  => 'WeibullDistribution[α, β]',
        to    => 'ExponentialDistribution[λ]',
        label => '\alpha = 1 \land \beta = \lambda^{-1}'
    },
    {
        from  => 'WeibullDistribution[α, β]',
        to    => 'FrechetDistribution[α, β]',
        label => 'X^{-1}'
    },
    {
        from  => 'WeibullDistribution[α, β]',
        to    => 'GumbelDistribution[a, b]',
        label => 'm + k\log(cX) \land c > 0 \land k > 0 \land a = m + k\log(c) + k\log(\beta) \land b = k\alpha^{-1}'
    },
    {
        from  => 'WeibullDistribution[α, β]',
        to    => 'ExtremeValueDistribution[a, b]',
        label => 'm - k\log(cX) \land c > 0 \land k > 0 \land a = m - k\log(c) - k\log(\beta) \land b = k\alpha^{-1}'
    },
    {
        from  => 'WeibullDistribution[α, β]',
        to    => 'RayleighDistribution[σ]',
        label => '\alpha = 2 \land \beta = \sqrt{2}\,\sigma'
    },
    {
        from  => 'ExponentialDistribution[λ]',
        to    => 'WeibullDistribution[α, β]',
        label => '\lambda = 1 \land X^{\alpha^{-1}}\beta'
    },
    {
        from  => 'WeibullDistribution[α, β, μ]',
        to    => 'WeibullDistribution[α, β]',
        label => '\mu = 0'
    },
    {
        from  => 'GammaDistribution[a, β, γ, μ]',
        to    => 'WeibullDistribution[α, β, μ]',
        label => 'a = 1 \land \gamma = \alpha'
    },
    {
        from  => 'UniformDistribution[{min, max}]',
        to    => 'WeibullDistribution[α, β]',
        label => '\min = 0 \land \max = 1 \land \beta\left(-\log(1-X)\right)^{\alpha^{-1}}'
    },
    {
        from  => 'MinStableDistribution[μ, σ, ξ]',
        to    => 'WeibullDistribution[α, β]',
        label => '\mu = \beta \land \sigma = \alpha^{-1}\beta \land \xi = -\alpha^{-1}'
    },
    {
        from  => 'MaxStableDistribution[μ, σ, ξ]',
        to    => 'WeibullDistribution[α, β]',
        label => '\mu = -\beta \land \sigma = \alpha^{-1}\beta \land \xi = -\alpha^{-1} \land -X'
    }
;
deduce-type(@edges)
```

```raku
my %edge-labels;
for @edges -> %r {
    %edge-labels{%r<from>} = {} unless %edge-labels{%r<from>}:exists;
    %edge-labels{%r<from>}{%r<to>} = %r<label>
}

deduce-type(%edge-labels):tally
```

```raku
&latex-interpret.signature
```

```raku
@edges.map(*<label>)[4]
```

```raku
.say for @edges.map(*<label>).map({ latex-interpret($_, actions => 'ascii-math') })
```

```raku
my $g = Graph.new(@edges):d
```

```raku
#% html
$g.dot(engine => 'fdp', 
    :edge-labels, :8graph-size, :6vertex-font-size, :1edge-thickness, arrow-size => 0.4,
    vertex-shape => 'rect', vertex-width => 1.2, vertex-height => 0.3,
    :%edge-labels, edge-font-size => 7
):svg
```
